```python
import redis
# example connection parameters
config = {
    "host": "redis-stack",
    "port": 6379,
    "decode_responses": True
}

# actual connection parameters are provided by the Notebook environment
r = redis.StrictRedis(**config)
r.info()
print ("Connected to DB-%d@%s:%d" % (r.connection_pool.connection_kwargs['db'], r.connection_pool.connection_kwargs['host'], r.connection_pool.connection_kwargs['port']))
```

```python
r.set('counter', 100);
r.incr('counter');
r.incrby('counter', 9);
r.decrby('counter', 4);
r.decr('counter');
r.incrbyfloat('counter',1.5);
print(r.get('counter'))
```

```python
r.delete('sessions:ggvd')

r.lpush('sessions:ggvd', 'Sesion02')
r.rpush('sessions:ggvd', *['Sesion04', 'Sesion05'])
r.lpush('sessions:ggvd', 'La Sesion 01')
r.rpop('sessions:ggvd')
r.linsert('sessions:ggvd', 'BEFORE', 'Sesion04', 'Sesion03')
r.lset('sessions:ggvd', 0, 'Sesion01')

response = r.lrange('sessions:ggvd', 0, -1)
print(response)
	
r.delete('sessions:ggvd');
```

```python
r.delete('students:ggvd')

r.sadd('students:ggvd', 'student1', 'student2', 'student3')
r.sadd('students:ggvd', 'student1')

message = r.smembers('students:ggvd')
print(message)

r.srem('students:ggvd', 'student3')

message = r.smembers('students:ggvd')
print(message)

message = r.scard('students:ggvd')
print(message)

r.delete('students:ggvd');
```

```python
r.delete('students:ggvd')

r.sadd('students:ggvd', 'student1', 'student2', 'student3')
r.sadd('students:bd', 'student3', 'student4', 'student5')

message = r.sunion('students:bd', 'students:ggvd')
print ('Unión: ', message)

message = r.sinter('students:bd', 'students:ggvd')
print ('Intersección: ', message)

message = r.sdiff('students:bd', 'students:ggvd')
print ('Diferencia: ', message)

r.delete('students:ggvd')
```

```python
r.delete('scores:ggvd')

r.zadd('scores:ggvd', {'student1': 9})
r.zadd('scores:ggvd', {'student2': 3})
r.zadd('scores:ggvd', {'student3': 8})

r.zincrby('scores:ggvd', 1, 'student2')

message = r.zcount('scores:ggvd', 5, 10)
print ('Cantidad de estudiantes aprobados: ', message)

message = r.zrangebyscore('scores:ggvd', 5, 10)
print ('Estudiantes que han aprobado: ', message)

r.delete('scores:ggvd');
```

```python
r.delete('profesor:mtorres');

r.hset('profesor:mtorres', 'email', 'mtorres@ual.es');
r.hset('profesor:mtorres', 'name', 'Manuel');
r.hset('profesor:mtorres', 'surname', 'Torres Gil');
r.hset('profesor:mtorres', 'twitter', '@ualmtorres');

message = r.hget('profesor:mtorres', 'email')
print ('Email: ', message)

message = r.hkeys('profesor:mtorres')
print ('Claves: ', message)

r.delete('profesor:mtorres');
```

```python
r.delete('a', 'b')

t = r.pipeline()   # pipeline() o pipeline(transaction=True) es el equivalente a MULTI
t.set('a', '1')
t.set('b', '2')
t.execute()        # execute() es el equivalente a EXEC

message = r.get('a')
print('a after exec: ', message)

message = r.get('b')
print('b after exec: ', message)

t = r.pipeline()
t.set('a', '3')
t.set('b', '4')
t.discard()

message = r.get('a')
print ('a after discard: ', message)

message = r.get('b')
print('b after discard: ', message)

r.delete('a', 'b')
```
